## Ejercicios

### 1- Crear una VIEW

Let's create a view called EMPSALARY to display salary along with some basic sensitive data of employees from the HR database. To create the EMPSALARY view from the EMPLOYEES table, Copy the code below and paste it to the textarea of the SQL page.

```python
%%sql
-- Primero, veo las columnas de la tabla employees
SELECT column_name, data_type, is_nullable
FROM information_schema.columns
WHERE table_name = 'employees';
```

```python
%%sql
-- Si la view existe la dropeo y la creo
DROP VIEW IF EXISTS empsalary;
CREATE VIEW empsalary AS
SELECT emp_id, f_name, l_name, b_date, sex, salary
FROM EMPLOYEES;
```

```python
#puedo tratar a la view como una tabla
%sql SELECT * FROM empsalary;
```

### 2- Update a View

Podemos crear o actualizar una view con info de otras tablas o otras views:

```python
%%sql
SELECT column_name, data_type, is_nullable
FROM information_schema.columns
WHERE table_name = 'jobs';
```

```python
%%sql
CREATE OR REPLACE VIEW empsalary AS
SELECT emp_id, f_name, l_name, b_date, sex, salary, min_salary, max_salary
FROM employees, jobs
WHERE employees.job_id = jobs.job_id;
```

```python
%%sql
SELECT * FROM empsalary;
```

Hicimos un implicit inner join, pero oteniendo una view como resultado en vez de una tabla

### 3- DROP a View

```python
%%sql
DROP VIEW empsalary;
```

```python
%%sql
SELECT * FROM empsalary;
```
