## VIEWS


Una view es como una tabla, pero no se guarda como tal. Una view accede a partes de una tabla o tablas, y te deja ver cierta informacion, pero sin crear una nueva tabla.  
Se puede crear una view usando otras views, incluso en combinacion con tablas y otras views.

Se puede pensar como una query nombrada, pues la forma en que construimos la view es dando nombre al resultado de una query.  
Las VIEWS son dinámicas: solo la definición de la view es almacenada, no la data.

````python
%sql SELECT * FROM employees;
````

````python
%%sql
SELECT column_name, data_type, is_nullable
FROM information_schema.columns
WHERE table_name = 'employees';
````

### Como crear una VIEW

Le damos un nombre al resultado de una query:
```sql
CREATE VIEW nombre_view AS
query;
```
Donde query puede ser cualquier result set. Este puede provenir de una sola tabla, de combinaciones de tablas, incluso de otras views

````python
%%sql
-- Ejemplo
DROP VIEW IF EXISTS empinfo;
CREATE VIEW empinfo (emp_id, f_name, l_name, address, job_id, manager_id, dep_id)
AS SELECT emp_id, f_name, l_name, address, job_id, manager_id, dep_id
FROM EMPLOYEES;
````

Podemos manipular la view como si fuera una tabla. A efectos practicos es una tabla, pero en realidad es una query que se corre cada vez que la llamamos, y nos da un result set que procede de la manipulacion de una o mas tablas o views.

````python
%sql SELECT * FROM empinfo;
````

Podemos ver que la view no aparece en la lista de tablas:

````python
%%sql
SELECT tablename, schemaname 
FROM pg_tables
WHERE schemaname NOT IN ('pg_catalog', 'information_schema');
````

Sin embargo, si aparece en la lista de views:

````python
%%sql
SELECT viewname, schemaname 
FROM pg_views
WHERE schemaname NOT IN ('pg_catalog', 'information_schema');
````

## STORES PROCEDURES

Son funciones, son un conjunto de comandos que viven en el servidor y que el usuario puede invocar con un solo comando.  
En vez de que el usuario mande una gran cantidad de comandos al servidor, simplemente manda uno y todos los comandos que estan en el servidor asociados a ese llamado de funcion, se corren.

Los stores procedures se pueden crear en diferentes lenguajes de programacion.

La diferencia entre un store procedure y una funcion es que la primera no devuelve resultados, y la segunda devuelve si o si un result set, una tabla (aunque sea una tabla de un elemento, una matriz de 1x1).

### Como crear un store procedure

Podemos tener muchos comandos adentro de un store procedure. Es necesario que haya otro delimitador, ademas del ;, que le diga a SQL donde terminan los comandos que estamos metiendo en el store procedure. Ahi es donde entra el AS \\$\\$.  
Podriamos usar otros delimitadores, por ejemplo @, o //, pero es convencion usar \\$\\$.  
Asi, todo lo que este entre \\$\\$ ... \\$\\$ va a ser considerado un solo bloque de instrucciones.  
Cuando corramos ese bloque de instrucciones en otro entorno que no sea este, sin embargo, el delimitador va a volver a ser ; por lo que nuestras queries se van a partir en las instrucciones deseadas.

````python
%%sql
DROP PROCEDURE IF EXISTS aumentar_salario_1;
CREATE PROCEDURE aumentar_salario_1(empNum INTEGER) 
LANGUAGE plpgsql 
AS $$
BEGIN
    UPDATE employees 
    SET salary = salary * 1.10
    WHERE emp_id = empNum;
END;
$$;
````

````python
%sql SELECT * FROM employees;
````

### Como llamar a un procedimiento

Usamos parentesis, tanto si le pasamos argumentos como si no.

````python
%%sql
UPDATE employees SET salary = 50000 WHERE emp_id = 2;
CALL aumentar_salario_1(2);
````

````python
%sql SELECT * FROM employees;
````

### Como ver los store procedures actuales

````python
%%sql
SELECT proname AS procedure_name, nspname AS schema_name, proargtypes AS parameter_types
FROM pg_proc
JOIN pg_namespace ON pg_proc.pronamespace = pg_namespace.oid
WHERE pg_proc.prokind = 'p';  -- 'p' indica PROCEDURE;
````

### Como eliminar un procedimiento

Dropeamos el store procedure sin olvidar el tipo de parametros que acepta.

````python
%%sql
DROP PROCEDURE IF EXISTS aumentar_salario_1(INTEGER);
````

## EXTRA: Funciones

Las funciones a diferencia de los store procedures, devuelven algo, result sets. No son void, como los store procedures, obligatoriamente deben devolver algo.

### Como crear una funcion

Las funciones pueden devolver un solo elemento (una matriz de 1x1):

````python
%%sql
-- EJ: Una funcion que devuelve el salario de un empleado dado su emp_id
DROP FUNCTION IF EXISTS obtener_salario(INTEGER);
CREATE FUNCTION obtener_salario(empleado_id INTEGER) 
RETURNS NUMERIC 
LANGUAGE plpgsql 
AS $$
DECLARE salario NUMERIC;
BEGIN
    SELECT salary INTO salario FROM employees WHERE empleado_id = emp_id;
    RETURN salario;
END;
$$;
````

O pueden devolver una tabla entera:

````python
%%sql
DROP FUNCTION IF EXISTS obtener_empleado(INTEGER);
CREATE FUNCTION obtener_empleado(empleado_id INTEGER) 
RETURNS TABLE(emp_id_n INT, f_name_n VARCHAR(50), salary_n NUMERIC) 
LANGUAGE plpgsql 
AS $$
BEGIN
    RETURN QUERY SELECT emp_id, f_name, salary FROM employees WHERE emp_id = empleado_id;
END;
$$;
````

### Como llamar a una funcion

Simplemente llamamos pero debemos SELECT el resultado, para que se vuelva un result set

````python
%%sql
SELECT obtener_salario(1);
````

````python
%%sql
SELECT * FROM obtener_empleado(1);
````
