## Ejemplos

### 1- Retrieve all

```python
%%sql
DROP PROCEDURE IF EXISTS obtener_perros;
DELIMITER $$
CREATE PROCEDURE obtener_perros()
BEGIN
    SELECT * FROM PETSALE
        WHERE type_animal = 'Dog';
    SELECT * FROM PETSALE
        WHERE id_animal = 1;
END $$
DELIMITER ;
```

```python
%%sql
CALL obtener_perros();
```

### 2- Update saleprice

```python
%%sql
DELIMITER @
    CREATE PROCEDURE UPDATE_SALEPRICE (IN Animal_ID INTEGER, IN Animal_Health VARCHAR(5))
    BEGIN
        IF Animal_Health = 'BAD' THEN
            UPDATE PETSALE
            SET SALEPRICE = SALEPRICE - (SALEPRICE * 0.25)
            WHERE ID = Animal_ID;
        ELSEIF Animal_Health = 'WORSE' THEN
            UPDATE PETSALE
            SET SALEPRICE = SALEPRICE - (SALEPRICE * 0.5)
            WHERE ID = Animal_ID;
        ELSE
            UPDATE PETSALE
            SET SALEPRICE = SALEPRICE
            WHERE ID = Animal_ID;
        END IF;
    END @
    DELIMITER ;
```

```python
%%sql
CALL UPDATE_SALEPRICE(2, 'BAD');
```
